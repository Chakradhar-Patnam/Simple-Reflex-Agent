```python
"""
Simple Reflex Agent Project: Two-Room Vacuum World

The agent selects an action only from its current percept.
It does not store history, learn from experience, or use an internal model.
"""


def simple_reflex_agent(percept):
    """Choose an action using condition-action rules only."""
    location, current_condition, all_clean = percept

    # Rule 1: If all rooms are currently clean, stop.
    if all_clean:
        return "Stop"

    # Rule 2: If the current room is dirty, clean it.
    if current_condition == "Dirty":
        return "Clean"

    # Rule 3: If Room A is clean, move to Room B.
    if location == "A":
        return "Move Right"

    # Rule 4: If Room B is clean, move to Room A.
    return "Move Left"


def run_environment(room_a, room_b, start_location, test_name, max_steps=10):
    """Simulate the environment and print each percept-action decision."""
    rooms = {"A": room_a, "B": room_b}
    location = start_location

    print("\n" + "=" * 60)
    print(test_name)
    print(f"Starting state: Room A={rooms['A']}, Room B={rooms['B']}, Agent={location}")
    print("=" * 60)

    for step in range(1, max_steps + 1):
        # The environment creates the CURRENT percept.
        # all_clean is current sensor information, not remembered history.
        all_clean = rooms["A"] == "Clean" and rooms["B"] == "Clean"
        percept = (location, rooms[location], all_clean)

        action = simple_reflex_agent(percept)
        print(f"Step {step}: Percept={percept} -> Action={action}")

        # The environment applies the selected action.
        if action == "Clean":
            rooms[location] = "Clean"
        elif action == "Move Right":
            location = "B"
        elif action == "Move Left":
            location = "A"
        elif action == "Stop":
            print("Agent: All rooms are clean. Task complete.")
            break
    else:
        print("Environment: Maximum step limit reached.")

    print(f"Final state: Room A={rooms['A']}, Room B={rooms['B']}, Agent={location}")
    return rooms, location


if __name__ == "__main__":
    # Test 1: Both rooms dirty, agent starts in Room A.
    run_environment("Dirty", "Dirty", "A", "TEST 1 - Both Rooms Dirty")

    # Test 2: Room A clean and Room B dirty, agent starts in Room A.
    run_environment("Clean", "Dirty", "A", "TEST 2 - One Room Dirty")

    # Test 3: Room A dirty and Room B clean, agent starts in Room B.
    run_environment("Dirty", "Clean", "B", "TEST 3 - Start in Different Location")

    # Optional Test 4: Both rooms already clean.
    run_environment("Clean", "Clean", "A", "TEST 4 - Both Rooms Already Clean")
```
